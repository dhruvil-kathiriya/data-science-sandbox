```python
import pandas as pd

data = {
    "UserID": [1, 2, 3, 4, 5, 6, 7],
    "CloudService": ["AWS", "Azure", "GCP", "AWS", "Azure", "AWS", "GCP"],
    "UsageHours": [120, 85, 140, 200, 95, 130, 180],
    "Cost": [300, 220, 350, 500, 250, 400, 450]
}

df = pd.DataFrame(data)
df.to_csv("cloud_data.csv", index=False)

df = pd.read_csv("cloud_data.csv")
df.head()
```

```python

df.describe()

usage_summary = df.groupby("CloudService")["UsageHours"].mean()
print(usage_summary)

import matplotlib.pyplot as plt
usage_summary.plot(kind="bar", title="Average Usage Hours per Cloud Service", color="skyblue")
plt.show()
```

**Calculate average sepal length for each species**

```python
df.groupby("species")["sepal_length"].mean()
```

**Make a simple scatter plot of sepal_length vs petal_length, coloured by species.**

```python
import matplotlib.pyplot as plt

for species in df["species"].unique():
    subset = df[df["species"] == species]
    plt.scatter(subset["sepal_length"], subset["petal_length"], label=species)

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Sepal vs Petal Length by Species")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "/content/netflix_titles.csv"
df = pd.read_csv(url)

df.head()
```

```python

print("\nShape of dataset:", df.shape)

print("\nColumn names:", df.columns.tolist())

print("\nSummary statistics:")
print(df.describe(include="all"))
```

```python

print("\nMovies vs TV Shows:")
print(df["type"].value_counts())

print("\nTop 5 countries:")
print(df["country"].value_counts().head(5))

print("\nReleases per year:")
print(df["release_year"].value_counts().head(1))

print("\nMost common rating:")
print(df["rating"].value_counts().head(1))
```

```python

df["type"].value_counts().plot(kind="bar", color=["skyblue", "lightgreen"], title="Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()

df["release_year"].value_counts().sort_index().plot(kind="line", marker="o", title="Netflix Releases per Year")
plt.xlabel("Year")
plt.ylabel("Number of Releases")
plt.show()
```
