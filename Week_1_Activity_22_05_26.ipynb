```python
import pandas as pd
import matplotlib.pyplot as plt

url = "/content/netflix_titles.csv"
df = pd.read_csv(url)

df.head(10)
```

```python
print("Shape of dataset:", df.shape)
```

```python
print("Column Names :",df.columns.to_list())
```

```python
print("Summary statistics:", df.describe(include="all"))
```

```python
print("Number of Movies vs Tv Shows:",df["type"].value_counts())
```

```python
print("Top 5 countries",df["country"].value_counts().head(5))
```

```python
print("Most Netflix releases",df["release_year"].value_counts().head(1))
```

```python
print(df["type"].value_counts().plot(kind="bar",color=["skyblue", "lightgreen"],title="Movies vs TV Shows"))
```

```python
print(df["release_year"].value_counts().plot(kind="line",marker="o",title="number of releases per year"))
```
